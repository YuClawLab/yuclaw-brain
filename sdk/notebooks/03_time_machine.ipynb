# 03 · Time-machine replay

Replay a ticker's composite signal as it would have looked on a past date. Only data with `available_as_of <= as_of` feeds the computation — a guarantee enforced by point-in-time DB filters in C6 / C8 / C9.

## Replay AMD on three dates

```python
import yuclaw_py
client = yuclaw_py.Client()
for d in ['2026-03-01', '2026-04-15', '2026-05-13']:
    snap = client.replay('AMD', d)
    print(f"{d}: {snap['label']:<15s}  score={snap['score']:+.3f}")
```

## What does each component contribute on a given date?

```python
snap = client.replay('AMD', '2026-03-01')
import pandas as pd
pd.Series(snap['components']).round(3).rename('score').to_frame()
```

Notice C6 (event impact) and C8 (cascade) are point-in-time exact — they query events with `available_as_of <= as_of`. The market components (C1, C3, C4, C5, C7) are approximations on historical dates because v2.3.0 currently caches only the latest market snapshot.
