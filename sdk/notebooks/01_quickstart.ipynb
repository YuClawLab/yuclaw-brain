## Install + import

`pip install yuclaw-py`. The cell below imports the SDK and prints the version.

```python
import yuclaw_py
yuclaw_py.__version__
```

## Connect

`source='postgres'` reads the local yuclaw_events database — the default v3.0 operator setup. `source='api'` will work once the hosted REST endpoint goes live (Day 11).

```python
client = yuclaw_py.Client(source='postgres', dsn='dbname=yuclaw_events')
client.source
```

## A composite signal

```python
sig = client.signal('NVDA')
print('label :', sig['label'])
print('score :', round(sig['score'], 3))
print('time  :', sig['signal_time'])
sig['components']
```

Note the `compliance` dict on every signal-bearing return:

```python
sig['compliance']
```

## Why?

`client.why(ticker)` returns the same signal plus the top evidence events that informed it.

```python
why = client.why('NVDA', n_evidence=3)
for ev in why['evidence']:
    print(f"  {ev['event_type']:<15s}  mag={ev['magnitude']:.2f}  {ev['raw_excerpt'][:70]}")
```
