v3.0's central differentiator is the **evidence layer**: every signal ties back to a source SEC filing or deterministic cascade. This notebook walks through how to inspect the raw events feeding a ticker.

## All events for a ticker

```python
import yuclaw_py
client = yuclaw_py.Client()
events = client.events('AMD', since='2026-04-01')
print(f'{len(events)} accepted events for AMD since 2026-04-01')
events[['event_type', 'magnitude', 'direction', 'available_as_of', 'cascade_depth']].head(10)
```

## Event-type mix

Form 4 insider trades dominate volume; 8-K material events are the rarer signal.

```python
events['event_type'].value_counts()
```

## SourceLock: every event has a verifiable source URL

```python
for _, row in events.head(3).iterrows():
    print(f"{row['event_type']:<14s}  {row['available_as_of']}")
    print(f"  excerpt: {row['raw_excerpt'][:90]}")
    print(f"  source : {row['source_url']}")
    print()
```

## Cascade vs primary

Cascade events carry `cascade_depth >= 1` — they're derived from a primary event on an upstream ticker via the supply-chain graph. Their magnitudes are intentionally smaller.

```python
events.groupby('cascade_depth').agg(
    n=('event_id', 'count'),
    mean_magnitude=('magnitude', 'mean'),
).round(3)
```
